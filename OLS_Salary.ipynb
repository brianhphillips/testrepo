```python
# Import libraries
import pandas as pd
import statsmodels.api as sm
import kagglehub
```

```python
# Download latest version
path = kagglehub.dataset_download("ishank2005/slary-data")

print("Path to dataset files:", path)

# Find the CSV file in the downloaded path and load it into a pandas DataFrame
import os
# Assuming the CSV file is named 'Salary_Data.csv' inside the downloaded directory
csv_file_path = os.path.join(path, "Salary_Data.csv")
df = pd.read_csv(csv_file_path)
print("Dataset loaded successfully into 'df' DataFrame.")
```

```python
# Define independent (X) and dependent (y) variables
X = df["YearsExperience"]
y = df["Salary"]
```

```python
# Add a constant (intercept) to the model
# The previous definition of X = ["YearsExperience"] was a list of strings,
# which caused a UFuncTypeError when sm.add_constant tried to perform numerical operations.
# X needs to be a pandas DataFrame or numpy array containing numerical data.
# Assuming 'df' (from the dataset loading step) is available, we redefine X here.
X = df[["YearsExperience"]] # Ensure X is a DataFrame for sm.add_constant
X = sm.add_constant(X)
```

```python
# Fit the OLS regression model
model = sm.OLS(y, X).fit()
```

```python
# Print regression results
print(model.summary())
```
